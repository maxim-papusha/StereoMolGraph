# Atom-mapped reaction SMILES to StereoCondensedReactionGraph

This notebook builds a condensed reaction graph and a stereo condensed reaction graph from a mapped reaction SMILES example from the heid-lab reaction database.

```python
import rdkit.Chem.rdChemReactions

from stereomolgraph.rdmol2graph import RDMol2StereoMolGraph
```

```python
reaction_smiles = (
    "[CH2:1]1[CH:2]=[CH:3][N:4]([CH3:5])[CH:6]=[C:7]1[C:8]([NH2:9])=[O:10]."
    "[CH3:11][O:12][C:13](=[O:14])[C:15]([CH3:16])=[O+:17][O-:18]>>"
    "[CH2:1]1[C@H:2]2[C@@H:3]([N:4]([CH3:5])[CH:6]=[C:7]1[C:8]([NH2:9])=[O:10])"
    "[O:18][O:17][C:15]2([C:13]([O:12][CH3:11])=[O:14])[CH3:16]"
)

converter = RDMol2StereoMolGraph()
rdrxn = rdkit.Chem.rdChemReactions.ReactionFromSmiles(reaction_smiles)

crg = converter.crg_from_rdrxn(rdrxn)
print(type(crg).__name__)
print("atoms:", crg.n_atoms)
print("formed bonds:", sorted(map(sorted, crg.get_formed_bonds())))
print("broken bonds:", sorted(map(sorted, crg.get_broken_bonds())))
```

```python
scrg = converter(rdrxn)
print(type(scrg).__name__)
print("atoms:", scrg.n_atoms)
print("formed bonds:", sorted(map(sorted, scrg.get_formed_bonds())))
print("broken bonds:", sorted(map(sorted, scrg.get_broken_bonds())))
print("atom stereo entries:", len(scrg.atom_stereo))
print("bond stereo entries:", len(scrg.bond_stereo))
```

The same constructor can be used for other atom-mapped reaction SMILES where reactant and product atom-map sets match exactly.
